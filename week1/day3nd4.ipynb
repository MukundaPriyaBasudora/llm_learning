Tokens:
Dividing of long input textual data into small words/subwords/characters is called tokenization.
during training time, the words or tokens are interpreted in such a way how well two tokens depends on each other or come together , based on it ..in the output time the output tokens are being generated.

tiktoken is OpenAI's tokenizer library. It converts text into tokens (small pieces of text) that language models process.

```python
import tiktoken

encoding = tiktoken.encoding_for_model("gpt-4.1-mini")

tokens = encoding.encode("Hi my name is Ed and I like banoffee pie")
```

```python
tokens
```

```python
for token in tokens:
    token_t=encoding.decode([token])
    print(token_t)
```

```python
gemini=OpenAI(base_url="https://generativelanguage.googleapis.com/v1beta/openai/",api_key=api_key)
```

```python
messages=[
    {"role":"system","content":"you are a naughty assistant"},
    {"role":"user","content":"Hii I am Mukunda"}
    ]
messages
```

```python
response=gemini.chat.completions.create(model="gemini-3.5-flash-lite",messages=messages)
response.choices[0].message.content
```
