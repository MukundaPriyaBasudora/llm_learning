BELOW ARE THE BASE URLS FOR DIFFERENT API KEYS

```python

anthropic_url = "https://api.anthropic.com/v1/"
gemini_url = "https://generativelanguage.googleapis.com/v1beta/openai/"
deepseek_url = "https://api.deepseek.com"
groq_url = "https://api.groq.com/openai/v1"
grok_url = "https://api.x.ai/v1"
openrouter_url = "https://openrouter.ai/api/v1"
ollama_url = "http://localhost:11434/v1"
```

```python
#connecting to gemini,openrouter,groq,ollama through openAI Client library
gemini=OpenAI(base_url=gemini_url,api_key=gemini_api_key)
openrouter=OpenAI(base_url=openrouter_url,api_key=openrouter_api_key)
groq=OpenAI(base_url=groq_url,api_key=groq_api_key)
ollama=OpenAI(base_url=ollama_url,api_key="ollama")



```

```python
message=[{
    "role":"user","content":"Explain the conecpt of feminism to a person who don't believe in feminisim,mock him for his narrow mind"
},]
```

```python
response=gemini.chat.completions.create(model="gemini-3.5-flash-lite",messages=message)
display(Markdown(response.choices[0].message.content))
```

```python
response=groq.chat.completions.create(model="openai/gpt-oss-120b",messages=message)
display(Markdown(response.choices[0].message.content))
```

```python
response=openrouter.chat.completions.create(model="openai/gpt-oss-120b",messages=message)
display(Markdown(response.choices[0].message.content))
```

```python
response=ollama.chat.completions.create(model="llama3.2",messages=message)
display(Markdown(response.choices[0].message.content))
```

Training vs Inference time scaling

Training time scaling: training time scaling is making the model smarter by increasing the model base size, no.of parameters and volume of training data using massive GPU clusters

Inference time scaling: adding the model extra processing power while user waits searching.

```python
easy_puzzle = [
    {"role": "user", "content": 
        "You toss 2 coins. One of them is heads. What's the probability the other is tails? Answer with the probability only."},
]
```

```python
response=gemini.chat.completions.create(model="gemini-3.5-flash-lite",messages=easy_puzzle,reasoning_effort="minimal")
display(Markdown(response.choices[0].message.content))
```

```python
response=gemini.chat.completions.create(model="gemini-3.5-flash-lite",messages=easy_puzzle,reasoning_effort="low")
display(Markdown(response.choices[0].message.content))
```

```python
hard = """
On a bookshelf, two volumes of Pushkin stand side by side: the first and the second.
The pages of each volume together have a thickness of 2 cm, and each cover is 2 mm thick.
A worm gnawed (perpendicular to the pages) from the first page of the first volume to the last page of the second volume.
What distance did it gnaw through?
"""
hard_puzzle = [
    {"role": "user", "content": hard}
]
```

```python
response=gemini.chat.completions.create(model="gemini-3.5-flash-lite",messages=hard_puzzle)
display(Markdown(response.choices[0].message.content))
```
