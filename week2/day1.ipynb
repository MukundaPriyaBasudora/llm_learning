Let's do a little fun conversation to make a convo bewteen two LLMS

```python
gemini_model="gemini-3.5-flash-lite"
ollama_model="llama3.2"
gemini_system = "You are a chatbot who is very feminist; \
you don't tolerate any kind of anti-feminism behaviour and always believe in feminism,you give tit for tat to who consider feminism as joke or insult it."

ollama_system = "You are a very polite, courteous anti-feminist. You don't agree with \
feminism concept the other feminist says,  If the other person is argumentative, \
you try to calm them down and keep putting out your point."
gemini_messages = ["Hi , what's up"]
ollama_messages = ["Hello, feminist"]
```

```python
def call_gemini():
    messages = [{"role": "system", "content": gemini_system}]
    for gem, oll in zip(gemini_messages, ollama_messages):
        messages.append({"role": "assistant", "content": gem})
        messages.append({"role": "user", "content": oll})
    response = gemini.chat.completions.create(model=gemini_model, messages=messages)
    return response.choices[0].message.content
```

```python
call_gemini()
```

```python
def call_ollama():
    messages = [{"role": "system", "content": ollama_system}]
    for gem, oll in zip(gemini_messages, ollama_messages):
        messages.append({"role": "user", "content": gem})
        messages.append({"role": "assistant", "content": oll})
    messages.append({"role": "user", "content": gemini_messages[-1]})
    response = ollama.chat.completions.create(model=ollama_model, messages=messages)
    return response.choices[0].message.content
```

```python
call_ollama()
```

```python
gemini_messages = ["Hi , what's up"]
ollama_messages = ["Hello, feminist"]

display(Markdown(f"### Gemini:\n{gemini_messages[0]}\n"))
display(Markdown(f"### Ollama:\n{ollama_messages[0]}\n"))

for i in range(5):
    gem_next = call_gemini()
    display(Markdown(f"### Gemini:\n{gem_next}\n"))
    gemini_messages.append(gem_next)
    
    ollama_next = call_ollama()
    display(Markdown(f"### Ollama:\n{ollama_next}\n"))
    ollama_messages.append(ollama_next)
```
